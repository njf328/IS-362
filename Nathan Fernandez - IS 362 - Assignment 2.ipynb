# Understanding the Python `zip()` Function

## Introduction

The `zip()` function is a useful Python function for working with multiple data structures at the same time. It can combine corresponding elements from two or more sequences, such as lists.

In this notebook, I will explore how the `zip()` function works, how it can be used in a `for` loop, and how it can be used to create a dictionary.

## What is `zip()`?

The `zip()` function combines elements from two or more sequences based on their position. For example, the first element from one list is paired with the first element from another list.

This can be useful when we want to work with related information stored in separate lists.

```python
names = ["Nathan", "Peter", "Bruce"]
scores = [96, 62, 39]

for name, score in zip (names, scores):
    print (name, score)
```

## How the Example Works

In this example, the `names` list contains names and the `scores` list contains scores. The `zip()` function pairs the elements according to their position.

The `for` loop then goes through each pair and prints the name followed by its corresponding score.

```python
products = ["Notebook", "Pen", "Folder"]
prices = [5, 1, 3]

for product, price in zip (products, prices):
    print (product, ": $", price)
```

## Using `zip()` With Related Data

The `zip()` function is useful when information is stored in separate lists, but the items in those lists are related.

In this example, each product is paired with its corresponding price. The `for` loop allows us to work with each pair one at a time.

```python
students = ["Kiryu", "Miles", "Clark"]
grades = [68, 11, 38]

student_grades = dict(zip(students, grades))

print (student_grades)
```

## Creating a Dictionary With `zip()`

In this example, `zip()` pairs each student with a grade. The `dict()` function then converts those pairs into a dictionary.

A dictionary is useful for storing information as key-value pairs. Here, each student's name is the key and the student's grade is the value.

```python
students = ["Victor", "Nishiki", "Lex"]
grades = [62, 68]

for student, grade in zip (students, grades):
    print (student, grade)
```

## What Happens With Lists of Different Lengths?

When using `zip()` with lists of different lengths, the resulting pairs stop when the shortest list runs out of elements.

In this example, there are three students but only two grades. Therefore, only the first two students are paired with grades.

This is important to remember when using `zip()` with multiple data structures.

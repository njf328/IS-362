## Scenario Design: Steam Customers

### 1. Who are your target users?

Steam's recommendation system primarily targets PC gamers who use the platform to discover, purchase, and play video games. These users include casual gamers, dedicated enthusiasts, and people who enjoy specific genres such as role-playing games, shooters, strategy games, and simulation games. Some customers already know what they want to play, while others rely on recommendations to discover something new.

### 2. What are their key goals?

The main goal for Steam customers is to find games that match their interests, preferences, and budgets without having to search through thousands of available titles. Customers may want to discover games similar to ones they already enjoy, explore unfamiliar genres, find highly rated games, or learn about new releases. They also want enough information to make informed purchasing decisions and avoid spending money on games they might not enjoy.

### 3. How can you help them accomplish those goals?

Steam can help customers accomplish these goals by using their gaming activity and preferences to recommend relevant titles. For example, Steam's store can suggest games associated with tags found in a customer's previously played games. Its Discovery Queue also introduces customers to games that they might not have searched for themselves. Displaying customer reviews, gameplay videos, genre tags, and other information alongside recommendations helps users evaluate whether a game is worth purchasing. Giving customers more control over their recommendations can further improve their experience by helping them discover games that better match their individual interests.

## Scenario Design: Valve's Perspective

Although Steam's recommendation system is designed to help customers discover games, it also serves Valve's business interests. Looking at the system from both perspectives helps explain why Steam might recommend certain games and how those recommendations could benefit the platform.

### 1. Who are your target users?

From Valve's perspective, the target users include Steam customers who browse the store, purchase games, and interact with personalized recommendations. The system also indirectly serves game developers and publishers who rely on Steam to help their games reach interested customers. Valve itself is another important stakeholder because it operates the platform and benefits when customers find games they want to purchase.

### 2. What are their key goals?

Valve's business goals likely include helping customers discover relevant games, increasing engagement with the Steam Store, encouraging purchases, and keeping customers satisfied with the platform. Another important goal is helping developers and publishers reach audiences who may be interested in their games. A successful recommendation system can make Steam more useful to customers while also supporting the platform's long-term growth.

### 3. How can you help them accomplish those goals?

Steam can support these goals by recommending games that customers are likely to find interesting based on their gaming activity, preferences, and other available signals. Relevant recommendations may encourage customers to explore more games and make purchases. Steam can also help developers reach potential customers by introducing players to titles they might otherwise overlook. From Valve's perspective, the recommendation system should balance commercial goals with customer satisfaction, since recommendations that consistently feel irrelevant or overly promotional could reduce users' trust in the platform.

## Reverse Engineering Steam's Recommendation System

Steam uses several recommendation features to help customers discover games, including its Featured & Recommended section, Discovery Queue, and Interactive Recommender. Although Valve does not publicly disclose every detail of its recommendation algorithms, examining these features provides clues about the information Steam uses to recommend games.

### Observations from My Steam Account

To better understand Steam's recommendation system, I examined the Steam Store and Discovery Queue while signed into my account. I found three examples that illustrate different reasons Steam gives for recommending games.

**Example 1: The Blood of Dawnwalker — Previously Played Games**

In the Featured & Recommended section, Steam recommended *The Blood of Dawnwalker* and explained that the recommendation was related to games I had previously played with the tags "Action-Adventure," "Third Person," and "Combat."

This suggests that Steam uses information about previously played games and their associated tags to identify other games that might match a customer's interests.

**Example 2: Melvor Idle 2: A RuneScape Adventure — Popularity**

While browsing my Discovery Queue, I encountered *Melvor Idle 2: A RuneScape Adventure*. Steam indicated that the game was included because it was popular.

This suggests that popularity can influence which games appear in the Discovery Queue, potentially introducing customers to titles outside their usual preferences.

**Example 3: Call of Duty: Modern Warfare 4 — Top Seller**

Another game in my Discovery Queue was *Call of Duty: Modern Warfare 4*, which Steam identified as a top seller. Interestingly, the game's displayed customer review sentiment was "Mixed."

This suggests that sales popularity and customer satisfaction are not necessarily the same thing. A game may be recommended because it sells well, even when customer reviews are divided.

### What These Observations Suggest

Based on these examples, Steam appears to use or highlight several different signals when presenting recommendations:

- **Gaming history and tags:** Games related to titles a customer has previously played.
- **Popularity:** Games that attract significant attention or activity.
- **Sales performance:** Games identified as top sellers.
- **Community reviews:** Ratings and review summaries that help customers evaluate recommended games.

These observations do not reveal the exact algorithms or the relative importance of each factor. They show the explanations Steam provides to customers and allow us to make reasonable inferences about its recommendation approach.

### Publicly Documented Recommendation Technology

Valve has also published information about its Interactive Recommender, a recommendation tool that uses machine learning to identify games a customer might enjoy.

According to Valve's explanation, the Interactive Recommender analyzes patterns in users' playtime histories to identify similarities between players and suggest games they have not yet played. Valve has also described controls that allow customers to adjust their recommendations based on factors such as a game's popularity and release date.

This approach resembles collaborative filtering because it uses similarities in user behavior to identify potentially relevant games.

It is important to distinguish the Interactive Recommender from Steam's other recommendation features. The explanations displayed in the Featured & Recommended section and Discovery Queue do not necessarily mean that those features use the exact same algorithm.

### Limitations of This Analysis

Because Steam's complete recommendation algorithms are proprietary, I cannot determine precisely how each recommendation is generated or how heavily different factors are weighted.

Additionally, these observations come from a single Steam account and a small number of recommendations. They provide useful examples of the user experience but are not enough to establish how the system behaves for all Steam customers.

## Recommendations for Improving Steam's Recommendation System

Based on my observations of Steam's recommendation features and the publicly available information about Valve's recommendation technology, I believe there are several ways Steam could improve how it helps customers discover games.

### 1. Provide More Detailed Recommendation Explanations

Steam already provides some explanations for its recommendations, such as identifying games as popular, top sellers, or similar to games a customer has previously played. However, explanations like "Popular" or "Top Seller" do not necessarily tell customers why a game would appeal to their individual interests.

I would recommend that Steam provide more detailed explanations whenever possible. For example, instead of only identifying a game as a top seller, Steam could also highlight relevant genres, gameplay features, or similarities to games the customer has enjoyed.

This could help customers better understand recommendations and decide whether a game is worth investigating.

### 2. Give Customers More Control Over Recommendation Priorities

Steam already offers some recommendation controls through its Interactive Recommender. However, I think similar controls could be more accessible throughout the main Steam Store and Discovery Queue.

For example, customers could choose to prioritize:

- Games similar to titles they have previously played.
- Games with especially positive customer reviews.
- New releases and trending games.
- Lesser-known independent games.
- Games within a preferred price range.

Allowing customers to adjust these priorities could make recommendations feel more personalized and useful.

### 3. Better Balance Popularity With Customer Satisfaction

One interesting observation from my Discovery Queue was that *Call of Duty: Modern Warfare 4* appeared because it was a top seller, even though its displayed customer reviews were "Mixed."

This does not necessarily mean the recommendation was inappropriate, but it illustrates why popularity and customer satisfaction should be considered separately.

I would recommend that Steam give customers an option to prioritize games with stronger review sentiment rather than primarily emphasizing popularity or sales performance.

This could help customers discover games that are both relevant to their interests and well received by other players.

### 4. Improve Discovery of Lesser-Known Games

Recommendation systems that emphasize popular games may make it harder for smaller or lesser-known titles to receive attention.

Although Steam's Interactive Recommender already provides controls related to popularity, I think Steam could make discovering lesser-known games easier by offering a clearly labeled discovery option focused on hidden gems.

This feature could recommend games that match a customer's interests but have relatively little visibility compared with major releases.

It could benefit customers by introducing them to games they might otherwise overlook, while also helping smaller developers reach interested audiences.

### Overall Recommendation

My main recommendation is for Steam to make its recommendation system more transparent and customizable across the entire store.

Steam already offers useful personalized recommendations, but giving customers clearer explanations and more control over what they see could improve the overall discovery experience.

These changes could help Steam balance its business interests with its customers' goals of finding enjoyable games.

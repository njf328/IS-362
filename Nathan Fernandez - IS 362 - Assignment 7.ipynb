## Introduction

The purpose of this assignment is to retrieve data from a web API using Python, read the returned JSON data, and transform it into a pandas DataFrame.

This assignment uses the New York Times Books API, which provides information about books appearing on The New York Times Best Sellers lists. The API allows users to retrieve book information such as titles, authors, rankings, and descriptions.

Python will be used to send a request to the Books API and process the JSON response. The pandas library will then be used to organize selected book information into a DataFrame for easier viewing and analysis.

This assignment demonstrates how web APIs can be used to retrieve structured data from an external source and present it in a tabular format.

### API Key Authentication

Access to the New York Times Books API requires an API key. For security purposes, the key is entered at runtime using Python's `getpass()` function rather than being stored directly in the notebook.

When running this notebook, users must enter their own New York Times API key when prompted. The key is not included in the submitted notebook or GitHub repository.

The notebook includes saved outputs from a successful API request so that the retrieved data and resulting DataFrame can be reviewed without rerunning the code.

```python
import requests
import pandas as pd
from getpass import getpass
```

```python
api_key = getpass("Enter your New York Times API key: ")
```

```python
url = "https://api.nytimes.com/svc/books/v3/lists/current/hardcover-fiction.json"

parameters = {
    "api-key": api_key
}

response = requests.get(url, params=parameters, timeout=30)

print ("HTTP status code:", response.status_code)
```

```python
data = response.json()

print ("API response status:", data["status"])
print ("Main JSON keys:", list(data.keys()))
print ("Number of books:", len(data["results"]["books"]))
```

```python
books = data["results"]["books"]

books_df = pd.DataFrame(books)

books_df = books_df[
    ["rank", "title", "author", "description", "weeks_on_list"]
]

books_df.head()
```

```python
print ("Number of books:", len(books_df))
print ("Number of columns:", len(books_df.columns))
print ("Column names:", list(books_df.columns))
```

## API Request and JSON Data

The New York Times Books API was used to retrieve information from the current Hardcover Fiction Best Sellers list. Python's `requests` library was used to send an HTTP GET request to the API endpoint.

An API key was required to authenticate the request. The key was entered securely using Python's `getpass()` function rather than being written directly into the notebook. This helps protect the API credentials when the notebook is shared.

The API returned an HTTP status code of **200**, indicating that the request was successful. The JSON response also contained a status of `OK`, confirming that the data was retrieved successfully.

The JSON response contained several main keys, including `status`, `copyright`, `num_results`, `last_modified`, and `results`. The book information was accessed through the `results` section using `data["results"]["books"]`.

The API returned **15 books** from the Hardcover Fiction Best Sellers list.

## DataFrame Creation and Results

The book records were converted into a pandas DataFrame using `pd.DataFrame()`. Five columns were selected to present the most relevant information:

1. **rank:** The book's position on the best-seller list.
2. **title:** The title of the book.
3. **author:** The name of the book's author.
4. **description:** A brief description of the book.
5. **weeks_on_list:** The number of weeks the book has appeared on the best-seller list.

The `books_df.head()` method was used to display the first five rows of the DataFrame while preserving all 15 records.

The completed DataFrame contained **15 rows and five columns**, demonstrating how JSON data retrieved from a web API can be transformed into a structured table using pandas.

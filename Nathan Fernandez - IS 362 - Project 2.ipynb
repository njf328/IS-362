## Dataset 1: Monthly Product Sales

This section analyzes a fictional dataset I created that shows monthly sales for three products at an online store.

The original dataset is organized in a wide format, with each month stored as a separate column. This structure is easy to read, but it is not ideal for some types of data analysis.

The goal of this section is to use Python and pandas to tidy the dataset by converting the monthly columns into a single Month variable and a Sales variable. The transformed data can then be used to analyze sales trends and compare the products over time.

```python
import pandas as pd

sales = pd.read_csv("Project 2 Dataset 1.csv")

sales
```

## Preparing the Data

The original dataset is organized in a wide format. Each product is represented by one row, while January through May are separate columns.

Although this format is easy to read, it is not ideal for analyzing sales trends across time. To make the dataset tidier, the month columns will be converted into values in a single `Month` column, and the corresponding sales amounts will be stored in a `Sales` column.

```python
tidy_sales = sales.melt(
    id_vars="Product",
    var_name="Month",
    value_name="Sales"
)

tidy_sales
```

### Tidy Data Transformation

The `melt()` function was used to transform the dataset from wide format into tidy format. The five month columns were converted into values in the `Month` column, while their corresponding sales values were placed in the `Sales` column.

The resulting dataset contains 15 observations, representing five months of sales for each of the three products. Each row now represents one product's sales for one month, which makes the data easier to analyze by product and over time.

```python
product_totals = tidy_sales.groupby("Product")["Sales"].sum()

product_totals
```

### Total Sales by Product

The total sales were calculated by grouping the tidy data by product and summing the sales values for each product.

The Wireless Mouse had the highest total sales with **405 units**. The Keyboard had **315 units**, while the Laptop Stand had **261 units**.

Based on these five months of data, the Wireless Mouse was the best-selling product overall.

```python
monthly_totals = tidy_sales.groupby("Month")["Sales"].sum()

monthly_totals
```

### Monthly Sales Trends

The monthly totals were calculated by grouping the tidy data by month and summing the sales for all three products.

April had the highest total sales with **228 units**, followed by May with **216 units**. January had the lowest total sales with **171 units**.

Overall, sales increased from January through February, decreased slightly in March, increased substantially in April, and then decreased slightly in May. This shows some variation in monthly sales, although the five-month dataset is too small to establish a longer-term seasonal pattern.

```python
highest_sale = tidy_sales.loc[tidy_sales["Sales"].idxmax()]

highest_sale
```

### Highest Individual Monthly Sale

The highest individual monthly sale was identified using the `idxmax()` function, which locates the row containing the maximum sales value.

The Wireless Mouse had the highest individual monthly sales figure, with **92 units sold in April**. This was the highest sales value for any product in a single month in the dataset.

```python
tidy_sales.info()
```

## Conclusion for Dataset 1

The original sales dataset was transformed from a wide format into a tidy format using pandas. The `melt()` function converted the month columns into a single `Month` variable and placed the corresponding sales amounts into a `Sales` variable.

The analysis showed that the **Wireless Mouse** had the highest total sales across the five months, with **405 units**. April had the highest overall monthly sales with **228 units** across all three products. The highest individual monthly sales figure was **92 units**, recorded for the Wireless Mouse in April.

The tidy structure makes it easier to analyze sales by product and month and provides a more useful format for further analysis.

## Dataset 3: NBA Player Statistics

This section analyzes a dataset created by classmate Elian Martinez for the Week 5 discussion. The dataset contains game statistics for four NBA players, including their team, points, rebounds, assists, and field goals made and attempted.

The original dataset is untidy because multiple variables are stored together in the `Game Stats` and `Shooting` columns. For example, the `Game Stats` column contains points, rebounds, and assists in a single field, while the `Shooting` column contains field goals made and attempted in a single field.

The goal of this section is to use Python and pandas to separate these combined values into individual variables. The transformed data can then be used to compare player performance and shooting statistics.

```python
nba_stats = pd.read_csv("Project 2 Dataset 3.csv")

nba_stats
```

## Preparing the Data

The original dataset contains multiple statistics combined within individual columns. The `Game Stats` column stores points, rebounds, and assists together, while the `Shooting` column stores field goals made and field goals attempted together.

To create a tidier dataset, these combined values will be separated into individual columns for `Points`, `Rebounds`, `Assists`, `FG Made`, and `FG Attempted`. This will make each statistic its own variable and make the data easier to analyze and compare across players.

```python
nba_tidy = nba_stats.copy()

nba_tidy[["Points", "Rebounds", "Assists"]] = (
    nba_tidy["Game Stats"]
    .str.extract(r"(\d+)\s*PTS,\s*(\d+)\s*REB,\s*(\d+)\s*AST")
    .astype(int)
)

nba_tidy
```

```python
nba_tidy[["FG Made", "FG Attempted"]] = (
    nba_tidy["Shooting"]
    .str.extract(r"(\d+)/(\d+)\s*FG")
    .astype(int)
)

nba_tidy
```

```python
nba_tidy = nba_tidy.drop(columns=["Game Stats", "Shooting"])

nba_tidy
```

### Tidy Data Transformation

The original dataset stored multiple statistics together in the `Game Stats` and `Shooting` columns. These combined values were separated into individual columns for `Points`, `Rebounds`, `Assists`, `FG Made`, and `FG Attempted`.

After extracting the individual statistics, the original `Game Stats` and `Shooting` columns were removed because their information was now represented by the new variables. The resulting dataset has one column for each variable, making it easier to compare and analyze the players' performance.

```python
nba_tidy["FG Percentage"] = (
    nba_tidy["FG Made"] / nba_tidy["FG Attempted"] * 100
).round(2)

nba_tidy
```

### Field Goal Percentage

Field goal percentage was calculated by dividing each player's field goals made by their field goals attempted and multiplying by 100.

Devin Carter had the highest field goal percentage at **55.00%**, followed by Marcus Reed at **52.94%**. Jordan Ellis and Andre Lewis both had field goal percentages of **50.00%**.

```python
stat_leaders = nba_tidy.loc[
    [
        nba_tidy["Points"].idxmax(),
        nba_tidy["Rebounds"].idxmax(),
        nba_tidy["Assists"].idxmax()
    ],
    ["Player", "Points", "Rebounds", "Assists"]
]

stat_leaders
```

### Player Performance

The player statistics show that Devin Carter led the group in two categories. He scored the most points with **29** and also recorded the most assists with **8**.

Jordan Ellis led the group in rebounds with **11**. These results show how the tidy dataset makes it easier to identify and compare the leaders in individual statistical categories.

```python
nba_tidy.info()
```

## Conclusion for Dataset 3

The original NBA player statistics dataset was transformed into a tidier format using pandas. The combined values in the `Game Stats` column were separated into individual variables for points, rebounds, and assists, while the `Shooting` column was separated into field goals made and field goals attempted. The original combined columns were then removed.

The analysis showed that **Devin Carter** led the group in points with **29** and assists with **8**. **Jordan Ellis** recorded the most rebounds with **11**. After calculating field goal percentage, Devin Carter also had the highest shooting percentage at **55.00%**.

The final dataset contains no missing values and stores each player statistic as its own variable. This tidy structure makes the data easier to compare and provides a more useful format for further analysis.

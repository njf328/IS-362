## Question 1: Northernmost Airport

The initial pandas analysis identified Dillant Hopkins Airport (EEN) as the northernmost airport because its latitude was listed as 72.270833 degrees north in the CSV file.

However, this result appears to be an outlier in the source data. Dillant Hopkins Airport is located in New Hampshire, and additional airport information confirms that its latitude is approximately 42.898 degrees north.

Therefore, I will treat the latitude value of 72.270833 for EEN as an apparent data error and correct it before determining the northernmost airport.

```python
airports.loc[airports["faa"] == "EEN", "lat"] = 42.898
```

```python
northernmost = airports.sort_values("lat", ascending=False).head(5)

print (northernmost[["faa", "name", "lat", "lon"]])
```

Based on the corrected dataset, BRW is the northernmost airport among the candidates examined. This result will be verified using additional airport information.

### Handling the Outlier

The initial analysis showed an unusually high wind speed of 1048.36058 mph at EWR at hour 3. This value was far outside the range of the other observations for the day.

I investigated the value by dividing it by 1.15078, the approximate conversion factor from knots to miles per hour. The result was approximately 911, suggesting that the value may have originated from a raw value of 911 that was converted to miles per hour. Because 1048.36058 mph is not a plausible weather observation, I treated this value as an outlier and excluded it from the comparison rather than replacing it with an invented value.

After excluding the outlier, the maximum wind speeds were:

- EWR: 21.86482 mph
- JFK: 20.71404 mph
- LGA: 23.01560 mph

Therefore, LaGuardia Airport (LGA) had the highest recorded wind speed among the three New York area airports on February 12, 2013.

### Answer to Question 3

After accounting for the outlier in the weather data, **LaGuardia Airport (LGA)** had the windiest weather among the three New York area airports on February 12, 2013.

Its highest recorded wind speed was **23.01560 mph**, compared with **21.86482 mph at EWR** and **20.71404 mph at JFK**.

Therefore, the answer to Question 3 is **LaGuardia Airport (LGA)**.

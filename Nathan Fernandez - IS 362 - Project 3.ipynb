## Introduction

The purpose of this project is to retrieve information from a relational SQL database and display the results in a pandas DataFrame. This project uses the Chinook sample database, which contains information about music tracks, albums, customers, and purchases.

SQLite was selected as the relational database management system (RDBMS). Python's built-in `sqlite3` library will be used to connect to the database, while pandas will be used to execute a SQL query and display the results. Because SQLite support is included with Python, no additional database middleware needs to be installed.

The SQL query will join five tables: `Customer`, `Invoice`, `InvoiceLine`, `Track`, and `Album`. The resulting DataFrame will contain four columns: `LastName` and `FirstName` from the `Customer` table, `Name` from the `Track` table, and `Title` from the `Album` table.

Each row will represent a purchased track and its corresponding album. The results will be sorted alphabetically by customer `LastName` and then `FirstName`, following the requirements and sample output provided in the assignment instructions.

```python
import sqlite3
import pandas as pd
from pathlib import Path

database_path = Path("Chinook_Sqlite.sqlite")

if database_path.is_file():
    connection = sqlite3.connect(database_path)
    print ("Successfully connected to the Chinook database!")
else:
    print ("Database file not found. Please check its location.")
```

## Database Connection

The Chinook sample database is stored in a SQLite file named `Chinook_Sqlite.sqlite`. This file is located in the same folder as the Jupyter Notebook.

```python
query = """
SELECT
    Customer.LastName,
    Customer.FirstName,
    Track.Name,
    Album.Title
FROM Customer
JOIN Invoice
    ON Customer.CustomerId = Invoice.CustomerId
JOIN InvoiceLine
    ON Invoice.InvoiceId = InvoiceLine.InvoiceId
JOIN Track
    ON InvoiceLine.TrackId = Track.TrackId
JOIN Album
    ON Track.AlbumId = Album.AlbumId
ORDER BY
    Customer.LastName,
    Customer.FirstName;
"""

purchases = pd.read_sql_query(query, connection)

purchases.head()
```

```python
print ("Number of purchase records:", len(purchases))
print ("Number of columns:", len(purchases.columns))
```

## SQL Query and DataFrame Results

The SQL query combines information from five tables in the Chinook database to identify customers and the music tracks they purchased.

The query begins with the `Customer` table, which provides each customer's last name and first name. It then uses four `JOIN` operations to connect the related tables:

1. **Customer → Invoice:** Matches each customer to their invoices using `CustomerId`.
2. **Invoice → InvoiceLine:** Identifies the individual purchased items on each invoice using `InvoiceId`.
3. **InvoiceLine → Track:** Retrieves the name of each purchased track using `TrackId`.
4. **Track → Album:** Retrieves the corresponding album title using `AlbumId`.

The `SELECT` statement retrieves four columns: `LastName`, `FirstName`, `Name` (track name), and `Title` (album title). The `ORDER BY` statement sorts the results alphabetically by customer last name and then first name.

The query returned **2,240 purchase records and four columns**. Each row represents a purchased track and its associated album. Customers can appear in multiple rows because they may have purchased multiple tracks.

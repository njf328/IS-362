## Data Preparation

The movie ratings were stored in a CSV file and loaded into a pandas DataFrame using `pd.read_csv()`. The dataset contains five fictional respondents and six movies, with ratings ranging from 1 to 5.

Using `.info()`, I confirmed that the DataFrame contains seven columns and five rows. Four movie ratings are missing because some respondents have not seen those movies. These missing values appear as `NaN` and will be excluded from average calculations.

No additional data cleaning is necessary because the ratings are already stored in separate columns and use appropriate numerical data types.

```python
user_averages = movie_ratings.set_index("Person").mean(axis=1)

user_averages
```

```python
movie_averages = movie_ratings.drop(columns="Person").mean()

movie_averages
```

## Average Ratings by Respondent and Movie

I calculated the average rating for each respondent and each movie using pandas `.mean()`. Missing ratings were automatically excluded from the calculations.

Among the five respondents, Bruce had the highest average rating at 4.60, while Josh had the lowest at 3.40. This suggests that some respondents may be more generous with their ratings than others.

Among the six movies, *The Odyssey* received the highest average rating at 4.60, while *Insidious: Out of the Further* received the lowest at 3.25.

These averages provide a useful starting point, but they do not account for differences in how individuals use the rating scale. Normalization may help us compare ratings more fairly by adjusting for each respondent's typical rating behavior.

```python
ratings_only = movie_ratings.set_index("Person")

normalized_ratings = ratings_only.sub(ratings_only.min(axis=1), axis=0).div(
    ratings_only.max(axis=1) - ratings_only.min(axis=1), axis=0
)

normalized_ratings
```

```python
normalized_user_averages = normalized_ratings.mean(axis=1)

normalized_user_averages
```

```python
normalized_movie_averages = normalized_ratings.mean()

normalized_movie_averages
```

## Normalized Movie Ratings

I used min-max normalization to transform each respondent's ratings onto a scale from 0 to 1. For each respondent, I subtracted their lowest rating from each rating and divided the result by the difference between their highest and lowest ratings.

This created a new DataFrame containing normalized ratings while preserving the missing values.

After normalization, Kiryu had the highest average normalized rating at approximately 0.611, followed closely by Bruce at 0.600. Jill had the lowest normalized average at 0.200, despite having one of the higher original averages.

Among the movies, *The Odyssey* remained the highest-rated at 0.600, while *Insidious: Out of the Further* remained the lowest at 0.000. However, *Forgotten Island* moved ahead of *Spider-Man: Brand New Day* after normalization.

These results demonstrate that normalization can change how respondents and movies compare by accounting for differences in each person's rating range.

## Advantages and Disadvantages of Normalized Ratings

### Advantages

One advantage of normalization is that it helps account for differences in how people use a rating scale. Some respondents may consistently give higher ratings, while others may be more critical. Min-max normalization adjusts each person's ratings based on their own minimum and maximum values, making their relative preferences easier to compare.

Normalization can also reveal patterns that are less obvious in the original ratings. For example, *Forgotten Island* ranked below *Spider-Man: Brand New Day* using the original averages but ranked above it after normalization.

### Disadvantages

One disadvantage is that normalized ratings lose some of their original meaning. A normalized rating of 0 does not necessarily mean someone disliked a movie. It only means that the movie received that person's lowest rating. For example, Jill's original ratings of 4 became 0 because 4 was her lowest rating.

Another disadvantage is that min-max normalization can exaggerate small differences. Bruce's original ratings ranged only from 4 to 5, but normalization transformed them into 0 and 1, making the differences appear more dramatic.

Normalization can also be problematic when someone gives every movie the same rating, since the difference between their maximum and minimum would be zero.

### Conclusion

Normalized ratings can be useful when comparing individual preferences, especially when respondents use rating scales differently. However, the original ratings are still valuable because they show the actual scores people assigned. Using both original and normalized ratings provides a more complete understanding of the data.

```python
standardized_ratings = ratings_only.sub(
    ratings_only.mean(axis=1), axis=0
).div(
    ratings_only.std(axis=1), axis=0
)

standardized_ratings
```

```python
standardized_user_averages = standardized_ratings.mean(axis=1)

standardized_user_averages
```

```python
standardized_movie_averages = standardized_ratings.mean()

standardized_movie_averages
```

## Standardized Movie Ratings (Extra Credit)

For the extra-credit portion of the assignment, I used z-score standardization to transform each respondent's ratings. This involved subtracting each respondent's average rating from their individual ratings and dividing by their standard deviation.

Unlike min-max normalization, which produces values between 0 and 1, standardization produces values that can be positive or negative. Positive values represent ratings above a respondent's personal average, while negative values represent ratings below that average.

After standardization, each respondent's average standardized rating was approximately zero, as expected.

Among the movies, *The Odyssey* had the highest average standardized rating at approximately 0.459, indicating that respondents generally rated it above their personal averages. *Insidious: Out of the Further* had the lowest average at approximately -1.017, indicating that respondents generally rated it below their personal averages.

These results demonstrate how standardization can help compare movie ratings while accounting for differences in respondents' average ratings and rating variability.
